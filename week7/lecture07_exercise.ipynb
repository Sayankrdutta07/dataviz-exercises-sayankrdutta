```python
# Task 1
fig1 = px.treemap(df_fossil, path=["Region","Country","Source"], values="TWh",
                  color="Source", color_discrete_map=source_colors,
                  title="Fossil energy by region, country and source",
                  template="plotly_white", height=760)
fig1.update_traces(texttemplate="%{label}<br>%{value:,.0f} TWh",
                   hovertemplate="<b>%{label}</b><br>%{value:,.1f} TWh<extra></extra>")
fig1.data[0].marker.colors = [
    (source_colors[label] if label in source_colors and parent else "#D9D9D9")
    for label,parent in zip(fig1.data[0].labels,fig1.data[0].parents)
]
fig1.show()
display(Image(data=base64.b64decode("")))
```

```python
# Task 2
low_country = df_low.groupby("Country",as_index=False)["TWh"].sum().sort_values("TWh",ascending=False)
low_country["Root"] = "Low-carbon"
fig2 = px.treemap(low_country,path=["Root","Country"],values="TWh",
                  color_discrete_sequence=[course_blue],title="Low-carbon energy by country",
                  template="plotly_white",height=650)
fig2.update_traces(texttemplate="%{label}<br>%{value:,.0f} TWh",
                   hovertemplate="<b>%{label}</b><br>%{value:,.1f} TWh<extra></extra>",
                   root_color="#D9D9D9")
fig2.show()
display(Image(data=base64.b64decode("")))

fig3 = px.bar(low_country.sort_values("TWh"),x="TWh",y="Country",orientation="h",
              text="TWh",color_discrete_sequence=[course_blue],
              title=f"{low_country.iloc[0]['Country']} leads low-carbon energy use",
              labels={"TWh":"Energy (TWh)","Country":""},template="plotly_white",height=850)
fig3.update_traces(texttemplate="%{text:,.0f}",textposition="outside",cliponaxis=False)
fig3.update_layout(showlegend=False,margin=dict(t=65,l=10,r=65,b=10))
fig3.show()
display(Image(data=base64.b64decode("")))
```

**Task 2 reflection:** The horizontal bar chart communicates the ranking more clearly because every bar shares a common baseline, making country-to-country comparisons easier than comparing rectangle areas.

```python
# Task 3
region_source = df_fossil.groupby(["Region","Source"])["TWh"].sum().unstack(fill_value=0)
for source in ["Coal","Oil","Natural Gas"]:
    if source not in region_source.columns:
        region_source[source] = 0
region_source = region_source[["Coal","Oil","Natural Gas"]]
region_share = region_source.div(region_source.sum(axis=1),axis=0).sort_values("Coal",ascending=False).reset_index()
fig4 = px.bar(region_share,x=["Coal","Oil","Natural Gas"],y="Region",orientation="h",barmode="stack",
              category_orders={"Region":region_share["Region"].tolist()[::-1]},
              color_discrete_map={"Coal":course_blue,"Oil":grey,"Natural Gas":dark_grey},
              labels={"value":"Share of regional fossil energy","Region":""},
              title=f"{region_share.iloc[0]['Region']} has the highest coal share ({region_share.iloc[0]['Coal']:.1%})",
              template="plotly_white",height=560)
fig4.update_layout(xaxis=dict(tickformat=".0%",range=[0,1]),
                   legend=dict(orientation="h",y=1.08,x=0),margin=dict(t=80,l=10,r=10,b=10))
fig4.show()
display(Image(data=base64.b64decode("")))
```

**Task 3 reflection:** The treemap shows where fossil-energy volume is concentrated across the region–country–source hierarchy. The 100% stacked bar identifies which region has the highest coal share within its fossil-energy mix.
